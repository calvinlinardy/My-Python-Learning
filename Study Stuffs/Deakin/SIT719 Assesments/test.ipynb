```python
import numpy as np
import matplotlib.pyplot as plt

# Define x_A range
x_A = np.linspace(0, 100, 200)

# Define constraints
x_B_orange = (5 - 6 * x_A) / 4  # 6x_A + 4x_B >= 5  -> x_B >= (5 - 6x_A) / 4
x_B_mango = (5 - 4 * x_A) / 8  # 4x_A + 8x_B >= 5  -> x_B >= (5 - 4x_A) / 8
x_B_lime = (6 - 2 * x_A) / 7  # 2x_A + 7x_B <= 6  -> x_B <= (6 - 2x_A) / 7
x_B_prod = 100 - x_A  # x_A + x_B = 100  -> x_B = 100 - x_A

# Determine valid region by taking the minimum of valid x_B values
x_B_valid = np.maximum(0, np.minimum.reduce([x_B_orange, x_B_mango, x_B_lime, x_B_prod]))

# Plot constraints
plt.figure(figsize=(8, 6))
plt.plot(x_A, x_B_orange, label=r'$6x_A + 4x_B \geq 5$ (Orange)', linestyle='dashed')
plt.plot(x_A, x_B_mango, label=r'$4x_A + 8x_B \geq 5$ (Mango)', linestyle='dashed')
plt.plot(x_A, x_B_lime, label=r'$2x_A + 7x_B \leq 6$ (Lime)', linestyle='dashed')
plt.plot(x_A, x_B_prod, label=r'$x_A + x_B = 100$', linestyle='solid')

# Fill feasible region
plt.fill_between(x_A, 0, x_B_valid, color='gray', alpha=0.3, label="Feasible Region")

# Labels and legend
plt.xlabel(r'$x_A$ (Liters of A)')
plt.ylabel(r'$x_B$ (Liters of B)')
plt.title('Feasible Region for Beverage Production')
plt.legend()
plt.ylim(0, 100)
plt.xlim(0, 100)
plt.grid(True)

# Show the graph
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sympy import symbols, Eq, solve

# Define x_A range
x_A = np.linspace(0, 3, 200)

# Define constraints
x_B_1 = (5 - 2 * x_A) / 7  # From 2x_A + 7x_B = 5
x_B_2 = (5 - 6 * x_A) / 4  # From 6x_A + 4x_B = 5

# Solve for intersection point
x_A_sym, x_B_sym = symbols('x_A x_B')
eq1 = Eq(2*x_A_sym + 7*x_B_sym, 5)
eq2 = Eq(6*x_A_sym + 4*x_B_sym, 5)
solution = solve((eq1, eq2), (x_A_sym, x_B_sym))
x_A_opt, x_B_opt = solution[x_A_sym], solution[x_B_sym]

# Plot constraints
plt.figure(figsize=(8, 6))
plt.plot(x_A, x_B_1, 'b-', label=r'$2x_A + 7x_B = 5$')
plt.plot(x_A, x_B_2, 'g-', label=r'$6x_A + 4x_B = 5$')

# Mark the optimal solution
plt.scatter([x_A_opt], [x_B_opt], color='red', zorder=3, label="Optimal Solution")

# Labels and legend
plt.xlabel(r'$x_A$ (Liters of A)')
plt.ylabel(r'$x_B$ (Liters of B)')
plt.title('Graphical Solution for Optimal Cost')
plt.legend()
plt.grid(True)

# Save the figure BEFORE plt.show()
plt.savefig("/Users/clcx/Downloads/Feasible_Region_for_Beverage_Production.png", dpi=300)

# Now show the plot
plt.show()
```

```python
import pandas as pd
from io import StringIO

# File path for the Alipay transactions
file_path = '/Users/clcx/Downloads/alipay_record_20250214_093716.csv'

# Read the raw content
with open(file_path, 'r', encoding='ISO-8859-1') as file:
    transaction_data = file.read()

# Decode it manually using GB18030
decoded_content = transaction_data.encode("ISO-8859-1").decode("gb18030")

# Print first 1000 characters for debugging
print(decoded_content[:1000])

# Convert the decoded content into a structured dataframe with automatic delimiter detection
df_transactions = pd.read_csv(StringIO(decoded_content), sep=None, engine='python', on_bad_lines='skip', header=None)

# Try alternative delimiters if needed
if df_transactions.isnull().sum().sum() > 0:  # If there are many NaN values
    df_transactions = pd.read_csv(StringIO(decoded_content), delimiter=',', engine='python', on_bad_lines='skip', header=None)
    if df_transactions.isnull().sum().sum() > 0:
        df_transactions = pd.read_csv(StringIO(decoded_content), delimiter=';', engine='python', on_bad_lines='skip', header=None)
    if df_transactions.isnull().sum().sum() > 0:
        df_transactions = pd.read_csv(StringIO(decoded_content), delimiter='\t', engine='python', on_bad_lines='skip', header=None)

# Drop completely empty columns
df_transactions.dropna(axis=1, how='all', inplace=True)

# Fill missing values with empty strings
df_transactions.fillna("", inplace=True)

# Trim extra spaces
df_transactions = df_transactions.applymap(lambda x: x.strip() if isinstance(x, str) else x)

# Display the cleaned transactions with correct Chinese characters
print(df_transactions.head())
```
