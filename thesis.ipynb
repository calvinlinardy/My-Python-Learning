```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

# Data
layers_removed = ['Original', 'CONV_4_2', 'CONV_4_1', 'CONV_3_5', 'CONV_3_4', 'CONV_3_3', 'CONV_3_2', 'CONV_3_1']
accuracy_constant_depth = [0.9321, 0.9354, 0.9363, 0.9272, 0.9250, 0.9100, 0.9050, 0.8920]
accuracy_stochastic_depth = [0.9543, 0.9575, 0.9594, 0.9474, 0.9390, 0.9380, 0.9270, 0.9210]

# Set the Seaborn color palette
sns.set_palette("pastel")

# Create Plots folder if it doesn't exist
if not os.path.exists('Plots'):
    os.makedirs('Plots')

# Set common parameters for all plots
bar_width = 0.35
bar_gap = 0.15

# Save the Accuracy plot
fig, ax = plt.subplots(figsize=(12, 6))
bar_width = 0.35
bar_gap = 0.15
bar1 = ax.bar([i - bar_width / 2 for i in range(len(layers_removed))], accuracy_constant_depth, width=bar_width, edgecolor='white', label='Constant Depth', alpha=0.7)
bar2 = ax.bar([i + bar_width / 2 for i in range(len(layers_removed))], accuracy_stochastic_depth, width=bar_width, edgecolor='white', label='Stochastic Depth', alpha=0.7)

# Set different colors for each bar
for i, (bar_cd, bar_sd) in enumerate(zip(bar1, bar2)):
    bar_cd.set_color(sns.color_palette("pastel")[i])
    bar_sd.set_color(sns.color_palette("pastel")[i])

# Set labels and title
ax.set_xticks(range(len(layers_removed)))
ax.set_xticklabels(layers_removed)
ax.set_ylabel('Accuracy')
ax.set_title('Accuracy for ResNet50 with Constant Depth and Stochastic Depth')

# Set legend
ax.legend()

# Save the plot into Plots folder as JPG
plt.savefig(os.path.join('Plots', 'accuracy_plot_removed_layers.jpg'), format='jpg')

# Show the plot
plt.show()
```
